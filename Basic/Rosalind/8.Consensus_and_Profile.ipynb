```python
dictA = {}
dictC = {}
dictG = {}
dictT = {}

with open("hehe.txt", "r") as FR:
    fr = FR.read()
    a = fr.split(">")
    for i in a[1:]:
        b = i.split("\n", 1)[1]
        for j in range(len(b)):
            print(b[j])
            if b[j] == "A":
                if j in dictA:
                    dictA[str(j)] += 1
                else:
                    dictA[str(j)] = 0
            elif b[j] == "C":
                if j in dictC:
                    dictC[str(j)] += 1
                else:
                    dictC[str(j)] = 0
            elif b[j] == "G":
                if j in dictG:
                    dictG[str(j)] += 1
                else:
                    dictG[str(j)] = 0
            else:
                if j in dictT:
                    dictT[str(j)] += 1
                else:
                    dictT[str(j)] = 0
string = ""
for k in range(len(b)):
    #print
    c = str(dictA[k])
    if dictC[k] > c:
        c = str(dictC[k])
    if dictG[k] > c:
        c = str(dictG[k])
    if dictT[k] > c:
        c = str(dictT[k])
    string += c
string
#print("A: ",dictA.values())
#print("C: ",dictA.values())
#print("G: ",dictA.values())
#print("T: ",dictA.values())
```
